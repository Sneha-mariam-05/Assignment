#Banking System Using Python


```python
class Banking_system:
  def __init__(self,account_num,pin,username,password,balance):
    self.account_num=account_num
    self.pin=pin
    self.username=username
    self.password=password
    self.balance=balance
  def login(self):
    print("1.Login using Account number and PIN")
    print("2.Login using Username and Password")
    choice=int(input("Enter the Choice:"))
    if choice==1 :
      account=input("Enter the Account number:")
      pin=input("Enter the PIN :")
      if(self.account_num==account and self.pin==pin):
        print("Login Sucessfully")
        self.menu()
      else :
        print("Invalid Credentials")
    elif  choice==2:
       username=input("Enter the Username:")
       password=input("Enter the password:")
       if(self.username==username and self.password==password):
         print("Login Suceessfully")
         self.menu()
       else:
        print("Invalid Credentials")
    else:
      print("Invalid Choice")
  def menu(self):
    while True:
     print("-----BANKING OPERATIONS-----")
     print("1.Deposit")
     print("2.Withdrawl")
     print("3.Check Balance")
     print("4.EXIT")
     choice=int(input("Enter the choice"))
     if choice==1:
       self.deposit()
     elif choice ==2:
       self.withdrawl()
     elif choice==3:
       print("Current Balance :",self.balance)
     elif choice ==4:
       print("THANK YOU")
       break
     else:
       print("INVALID OPERATION")
  def deposit(self):
    amount=float(input("Enter the amount to be deposited:"))
    if amount<=0:
      print("Amount should be greater than zero")
    else:
      self.balance=self.balance+amount
      print("Amount successfully deposited")
      print("Current Balance:",self.balance)
  def withdrawl(self):
    amount=float(input("Enter the withdrawl amount:"))
    if amount<0:
      print("Amount must be greater than zero")
    elif amount>self.balance:
      print("Amount should be less than the available balance")
    else:
       self.balance=self.balance-amount
       print("Withdrwal successfull")
       print("Remaining balance:",self.balance)
bank =Banking_system("123456789","2008","Sneha","sneha@123",5000)
bank.login()


```
